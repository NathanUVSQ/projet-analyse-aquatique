Entrainement YOLO

```python
import os
from ultralytics import YOLO

# ----------------------------
# CONFIGURATION
# ----------------------------

EPOCHS = 100
IMAGE_SIZE = 224
BATCH_SIZE = 16

yaml_path = os.path.join(os.getcwd(), "dataset")

# ----------------------------
# ENTRAÎNEMENT
# ----------------------------

model = YOLO("yolo11n-cls.pt")

print("\n[INFO] Début de l'entraînement...\n")

model.train(
    data="dataset",
    epochs=EPOCHS,
    imgsz=IMAGE_SIZE,
    batch=BATCH_SIZE,
    freeze=5,
    patience=20,
    augment=True,
    hsv_h=0.0,
    hsv_s=0.2,
    hsv_v=0.3,
    fliplr=0.5,
    flipud=0.5,
    degrees=180,
    name="microorganismes_v1",
    exist_ok=True
)

# ----------------------------
# VALIDATION FINALE
# ----------------------------

print("\n[INFO] Validation...")
metrics = model.val()

print(f"\n[RÉSULTATS]")
print(f"  Accuracy top-1 : {metrics.top1:.2%}")
print(f"  Accuracy top-5 : {metrics.top5:.2%}")
print(f"\n[INFO] Modèle sauvegardé dans : runs/classify/microorganismes_v1/weights/best.pt")
```
